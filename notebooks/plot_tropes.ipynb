```python
import re
import json
from bs4 import BeautifulSoup
import pandas as pd

file_name = "plot_tropes.csv"
soup = BeautifulSoup(html_data, 'html.parser')
links = soup.find_all('a', class_='twikilink')
tropes_data = []
for link in links:
    url = link.get('href')
    # Extract the title from the last part of the URL
    title_match = re.search(r'/Main/([^/]+)$', url)
    if title_match:
        title = title_match.group(1)
        trope_info = {
            "trope": title,
            "url": url
        }
        tropes_data.append(trope_info)

df = pd.DataFrame(tropes_data)
df.to_csv(file_name, index=False)


print(f"Extracted {len(tropes_data)} tropes to plot_tropes.csv")
```

```python
from pathlib import Path
plot_tropes = pd.read_csv(file_name)
tvtropes_dataset_folder = Path("../data/tvtropes")
tropes = pd.read_csv(tvtropes_dataset_folder / "tropes.csv", index_col=0)
lit_tropes_examples = pd.read_csv(tvtropes_dataset_folder / "lit_goodreads_match.csv", index_col=0)

print(plot_tropes.columns.to_list())
print(tropes.columns.to_list())
print(lit_tropes_examples.columns.to_list())
```

```python
filtered_short_examples = tropes_examples_matched_with_books[tropes_examples_matched_with_books["Example"].map(lambda x: len(x.split(" ")) > 20)]
filtered_short_examples.to_csv("filtered_short_examples.csv", index=False)
print(len(filtered_short_examples))
filtered_short_examples.sample(10)
print(filtered_short_examples.loc[60])
```

```python
import numpy as np
filtered_short_examples = pd.read_csv("filtered_short_examples.csv")
save_file = Path("filtered_short_examples.npy")
# if True:
if not save_file.is_file():
    from FlagEmbedding import BGEM3FlagModel
    model = BGEM3FlagModel('BAAI/bge-m3', use_fp16=True)
    tokenizer = model.tokenizer
    embeddings = model.encode(filtered_short_examples["Example"].map(lambda x: x.strip()).to_list())
    embeddings = embeddings["dense_vecs"]
    with open('filtered_short_examples.npy', 'wb') as f:
        np.save(f, embeddings)
else:
    with open(save_file, "rb") as f:
        embeddings = np.load(f)
print(embeddings.shape)
```

```python
import hdbscan
import umap

umap_embeddings = umap.UMAP(n_neighbors=7, n_components=100, min_dist=0.0, metric='cosine').fit_transform(embeddings)
hdbscan_model = hdbscan.HDBSCAN(min_cluster_size=5, cluster_selection_method='leaf')
cluster_labels = hdbscan_model.fit_predict(umap_embeddings)

filtered_short_examples["cluster"] = cluster_labels



```

```python
import numpy as np
import pandas as pd
from collections import defaultdict
from sklearn.model_selection import train_test_split

# 1. Group examples by Trope
def prepare_nudge_dataset(df, embeddings):
    # Group by Trope and filter to keep only tropes with at least 10 examples
    trope_groups = defaultdict(list)
    trope_indices = defaultdict(list)
    
    for idx, row in df.iterrows():
        trope = row['Trope']
        trope_groups[trope].append(idx)
    
    # Filter to keep only tropes with at least 10 examples
    valid_tropes = {trope: indices for trope, indices in trope_groups.items() if len(indices) >= 10}
    
    # 2. Create query sets and positive documents
    queries = []
    query_embeddings = []
    query_ans_idx = []
    
    for trope, indices in valid_tropes.items():
        # For each trope, select one example as query
        for query_idx in indices:
            # The query is the example text
            query = df.loc[query_idx, 'Example']
            queries.append(query)
            
            # Get the embedding for this query
            query_embeddings.append(embeddings[query_idx])
            
            # All other examples with the same trope are positive documents
            positive_docs = [idx for idx in indices if idx != query_idx]
            query_ans_idx.append(positive_docs)
    
    # 3. Split into training and validation sets
    train_indices, val_indices = train_test_split(range(len(queries)), test_size=0.2, random_state=42)
    
    train_q_embs = np.array([query_embeddings[i] for i in train_indices])
    train_q_ans_indx = [query_ans_idx[i] for i in train_indices]
    
    val_q_embs = np.array([query_embeddings[i] for i in val_indices])
    val_q_ans_indx = [query_ans_idx[i] for i in val_indices]
    
    # 4. Prepare the final dataset format
    train_set = {'q_embs': train_q_embs, 'q_ans_indx': train_q_ans_indx}
    val_set = {'q_embs': val_q_embs, 'q_ans_indx': val_q_ans_indx}
    
    return train_set, val_set, embeddings

# Assuming filtered_short_examples is your DataFrame and example_embeddings is your numpy array of embeddings
train_set, val_set, data_embs = prepare_nudge_dataset(filtered_short_examples, embeddings)
# print(filtered_short_examples.loc[train_set['q_ans_indx'][0]])
# Now you can use these with NUDGE
from nudge import NUDGEN

# Fine-tune embeddings using NUDGE-N
finetuned_embs_nudge_n = NUDGEN().finetune_embeddings(data_embs, train_set, val_set)
```

```python
nudge_n_umap = umap.UMAP(n_neighbors=6, n_components=100, min_dist=0.0, metric='cosine').fit_transform(finetuned_embs_nudge_n)
hdbscan_model = hdbscan.HDBSCAN(min_cluster_size=5, cluster_selection_method='eom')
cluster_labels = hdbscan_model.fit_predict(nudge_n_umap)
filtered_short_examples["cluster"] = cluster_labels
filtered_short_examples["cluster_prob"] = hdbscan_model.probabilities_
```

```python
from FlagEmbedding import BGEM3FlagModel
model = BGEM3FlagModel('BAAI/bge-m3', use_fp16=True)
# question = "I'm looking for a story about a young woman who is mistreated by her stepfamily, forced to do household chores, and eventually finds love and escapes her difficult circumstances with the help of magical elements."
question = "I'm looking for a horror movie where a group of teenagers goes camping in the woods and they're hunted one by one by a supernatural entity or killer. The protagonist discovers a dark secret about the location's history that explains the killings."
question_embedding = model.encode(question)
question_embedding = question_embedding["dense_vecs"]

query_embeddings = np.array([question_embedding])

# find most similar examples to the question embedding
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

def get_topk_similar(query_embeddings, data_embeddings, k=5):
    similarities = cosine_similarity(query_embeddings, data_embeddings)
    topk_indices = np.argsort(-similarities, axis=1)[:, :k]
    topk_scores = np.take_along_axis(similarities, topk_indices, axis=1)
    return topk_indices, topk_scores

topk = 20
indices, scores = get_topk_similar(query_embeddings, data_embs, k=topk)

print(f"Top {topk} most similar examples to the question: '{question}'")
for qid in range(query_embeddings.shape[0]):
    for i in range(topk):
        idx = indices[qid][i]
        score = scores[qid][i]
        print(f"\nRank {i+1} (similarity: {score:.4f}):")
        print(f"Trope: {filtered_short_examples['Trope'].iloc[idx]}")
        print(f"Title: {filtered_short_examples['Title'].iloc[idx]}")
        print(f"Example: {filtered_short_examples['Example'].iloc[idx]}")






```
